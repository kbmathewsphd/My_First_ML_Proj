```python
import random
CHOICES = ("rock", "paper", "scissors")

def get_user_choice():
  while True:
    choice = input("Choose rock, paper, or scissors: ").strip().lower()
    if choice in CHOICES:
      return choice
    print("Invalid choice. Please type: rock, paper, or scissors.")

def decide_winner(user, computer):
  if user == computer:
    return "tie"
  if (user == "rock" and computer == "scissors") or (user == "paper" and computer == "rock") or (user == "scissors" and computer == "paper"):
    return "user"
    return "computer"

def main():
  user = get_user_choice()
  computer = random.choice(CHOICES)
  print("Computer chose:", computer)

  winner = decide_winner(user, computer)
  if winner == "tie":
    print("It's a tie!")
  elif winner == "user":
    print("You win!")
  else:
    print("Computer wins!")

if __name__ == "__main__":
  main()
```

```python
def format_pretty_10x(number, decimals=2):
    # Map for superscript characters
    superscripts = str.maketrans("0123456789-", "⁰¹²³⁴⁵⁶⁷⁸⁹⁻")

    scientific_str = f"{number:.{decimals}e}"
    base, exponent = scientific_str.split("e")

    # Clean the exponent and convert it to superscript characters
    clean_exponent = str(int(exponent)).translate(superscripts)

    return f"{base} × 10{clean_exponent}"

n1 = float(input("Enter a large number: "))
n2 = float(input("Enter a decimel number: "))

# Examples:
print(format_pretty_10x(n1))            # Output: 1.50 × 10⁴
print(format_pretty_10x(n2, 2))         # Output: 7.34 × 10⁻⁴
```

```python
def format_10x(number, decimals=2):
    # Format to native scientific notation first (e.g., "1.50e+04")
    scientific_str = f"{number:.{decimals}e}"

    # Split the coefficient and the exponent
    base, exponent = scientific_str.split("e")

    # Convert exponent to integer to remove leading zeros and signs (like +04 to 4)
    exponent = int(exponent)

    return f"{base} * 10^{exponent}"

n1 = float(input("Enter a large number: "))
n2 = float(input("Enter a decimel number: "))

# Examples:
print(format_10x(n1))       # Output: 1.50 * 10^4
print(format_10x(n2, 3)) # Output: 7.340 * 10^-4
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

np.random.seed(42)
X = np.random.rand(50, 1) * 100
Y = 3.5 * X + np.random.randn(50, 1) * 20

model = LinearRegression()
model.fit(X, Y)

Y_pred = model.predict(X)

plt.figure(figsize=(8,6))
plt.scatter(X, Y, color='blue', label='Data Points')
plt.plot(X, Y_pred, color='red', linewidth=2, label='Regression Line')
plt.title('Linear Regression on Random Dataset')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()

print("Slope (Coefficient):", model.coef_[0][0])
print("Intercept:", model.intercept_[0])
```

```python
import numpy as np
import matplotlib.pyplot as plt

def generate_primes(limit):
    sieve = np.ones(limit + 1, dtype=bool)
    sieve[:2] = False

    for i in range(2, int(limit**0.5) + 1):
        if sieve[i]:
            sieve[i*i : limit+1 : i] = False

    return np.nonzero(sieve)[0]

primes = generate_primes(50000)

r = primes
theta = primes  # radians

fig, ax = plt.subplots(figsize=(10, 10), subplot_kw={'projection': 'polar'})
ax.scatter(theta, r, s=2, c='red')
ax.set_title("Prime Number Spiral (r=n, θ=n)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Function to generate primes up to a limit using Sieve of Eratosthenes
def generate_primes(limit):
    sieve = np.ones(limit + 1, dtype=bool)
    sieve[:2] = False

    for i in range(2, int(limit**0.5) + 1):
        if sieve[i]:
            sieve[i*i : limit+1 : i] = False

    return np.nonzero(sieve)[0]

# Generate prime numbers up to 50,000
primes = generate_primes(50000)

# Distance = n
r = primes

# Angle = n degrees converted to radians
theta = np.deg2rad(primes % 360)

# Create polar plot
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='polar')

ax.scatter(theta, r, s=2, color='blue', alpha=0.7)

ax.set_title("Polar Plot of Prime Numbers (Distance=n, Angle=n°)", pad=20)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

N = 50000

# Sieve
prime = np.ones(N + 1, dtype=bool)
prime[:2] = False

for i in range(2, int(np.sqrt(N)) + 1):
    if prime[i]:
        prime[i*i:N+1:i] = False

# Spiral coordinates
x = y = 0
dx, dy = 0, -1

size = int(np.ceil(np.sqrt(N)))
coords = {}

for n in range(1, N + 1):
    coords[n] = (x, y)

    if (x == y) or (x < 0 and x == -y) or (x > 0 and x == 1 - y):
        dx, dy = -dy, dx

    x += dx
    y += dy

# Plot primes only
px = []
py = []

for n in range(2, N + 1):
    if prime[n]:
        x, y = coords[n]
        px.append(x)
        py.append(y)

plt.figure(figsize=(10, 10))
plt.scatter(px, py, s=1, color='black')
plt.axis('equal')
plt.axis('off')
plt.title("Ulam Spiral of Primes up to 50,000")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(math.sqrt(num)) + 1):
        if num % i == 0:
            return False
    return True

x = []
y = []

for n in range(100):
    value = n*n + n + 41
    if is_prime(value):
        x.append(n)
        y.append(value)

plt.figure(figsize=(10, 6))
plt.scatter(x, y, color='red')
plt.grid(True)
plt.xlabel("n")
plt.ylabel("n² + n + 41")
plt.title("Prime Values Generated by Euler's Polynomial")
plt.show()
```

```python
import math

def is_prime(num):
    if num < 2:
        return False
    if num == 2:
        return True
    if num % 2 == 0:
        return False

    limit = int(math.sqrt(num)) + 1
    for i in range(3, limit, 2):
        if num % i == 0:
            return False
    return True

# Euler's polynomial
def euler_polynomial(n):
    return n*n + n + 41

print(f"{'n':>3} {'f(n)':>6} {'Prime?':>8}")
print("-" * 20)

for n in range(40):
    value = euler_polynomial(n)
    print(f"{n:>3} {value:>6} {str(is_prime(value)):>8}")
```
